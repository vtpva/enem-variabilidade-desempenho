Este notebook conecta ao armazenamento de dados, inspeciona os arquivos parquet do ENEM e exibe as primeiras linhas da base ().

```python
import os
from google.colab import drive

# Mount Google Drive if not already mounted
if not os.path.exists('/content/drive/MyDrive'):
    drive.mount('/content/drive')
```

```python
import pandas as pd

tcc_dir = '/content/drive/MyDrive/TCC'
files = sorted([f for f in os.listdir(tcc_dir) if f.endswith('.parquet')])
print('Available parquet files:', files)

# Read one of the datasets
selected_file = 'MICRODADOS_ENEM_2022.parquet'
file_path = os.path.join(tcc_dir, selected_file)
print(f'\nLoading {file_path}...')

df = pd.read_parquet(file_path)
print(f'Successfully loaded: {df.shape[0]:,} rows x {df.shape[1]:,} columns\n')

df.head()
```
